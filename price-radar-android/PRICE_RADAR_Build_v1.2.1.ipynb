# PRICE RADAR Android 1.2.1

Нативное приложение: поиск внутри WebView по пяти магазинам и предварительное считывание видимых карточек. В версии 1.2.1 цены не учитываются в сравнении, пока пользователь не проверит прямую HTTPS-ссылку и цену в карточке и не нажмёт «Подтвердить».

Загрузите ZIP исходников. Тетрадь запускает тесты, затем сборку debug APK в Colab.


```python
from google.colab import files
import os, zipfile, shutil, subprocess

uploaded = files.upload()
zip_name = next((name for name in uploaded if name.lower().endswith('.zip')), None)
if not zip_name:
    raise RuntimeError('Загрузите PRICE_RADAR_ANDROID_SOURCE_v1.2.1.zip')

base = '/content/price-radar-android'
shutil.rmtree(base, ignore_errors=True)
with zipfile.ZipFile(zip_name) as archive:
    archive.extractall('/content')
if not os.path.isfile(base + '/build_in_colab.sh'):
    raise RuntimeError('В ZIP не найден build_in_colab.sh')
subprocess.run(['node', base + '/tests_store_parser.js'], check=True)
subprocess.run(['node', base + '/tests_offer_confirmation.js'], check=True)
subprocess.run(['bash', base + '/build_in_colab.sh'], check=True)
files.download(base + '/PRICE_RADAR_1.2.1_debug.apk')
```
