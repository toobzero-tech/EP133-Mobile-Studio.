# PRICE RADAR Android 1.2.0

Нативное приложение с поиском внутри WebView по DNS, М.Видео, Ozon, Wildberries и Яндекс Маркет. В этой версии добавлен ручной запуск считывания видимых карточек: распознанная цена и прямая ссылка импортируются в сравнение как предварительные. Проверка цены/наличия на карточке обязательна.

Загрузите ZIP исходников. Тетрадь проверит парсер на тестовых примерах, затем соберёт debug APK в Google Colab.


```python
from google.colab import files
import os, zipfile, shutil, subprocess

uploaded = files.upload()
zip_name = next((name for name in uploaded if name.lower().endswith('.zip')), None)
if not zip_name:
    raise RuntimeError('Загрузите PRICE_RADAR_ANDROID_SOURCE_v1.2.0.zip')

base = '/content/price_radar_android'
shutil.rmtree(base, ignore_errors=True)
with zipfile.ZipFile(zip_name) as archive:
    archive.extractall('/content')
if not os.path.isfile(base + '/build_in_colab.sh'):
    raise RuntimeError('В архиве не найден build_in_colab.sh')
subprocess.run(['bash', base + '/build_in_colab.sh'], check=True)
files.download(base + '/PRICE_RADAR_1.2.0_debug.apk')
```
